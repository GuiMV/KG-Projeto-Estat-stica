```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
#import numpy as np

df = pd.read_csv('https://raw.githubusercontent.com/GuiMV/KG-Projeto-Estat-stica/refs/heads/main/dataset/pokemon_data.csv')
amostra = pd.read_csv('https://raw.githubusercontent.com/GuiMV/KG-Projeto-Estat-stica/refs/heads/main/dataset/pokemon_amostra.csv')
```

# Funções

```python
def exibir_colunas(colunas, colunas_por_linha=6, largura_fixa=22):
    for i, col in enumerate(colunas, 1):
        print(col.ljust(largura_fixa), end='')
        if i % colunas_por_linha == 0:
            print()
    print()

def mostrar_dataset():
    pd.set_option('display.max_columns', 5)
    print(df.head(5))

def tamanho_dataset():
  print(f"O dataset possui {df.shape[0]} linhas e {df.shape[1]} colunas.")
  print(f"A amostra possui {amostra.shape[0]} linhas e {amostra.shape[1]} colunas.")

def mostrar_colunas():
  exibir_colunas(df.columns, colunas_por_linha=6)

def variaveis_qualitativas():
  colunas = ['ID', 'Name', 'Type 1', 'Type 2', 'Classification_info', 'gen']
  exibir_colunas(colunas, colunas_por_linha=2)

def variaveis_quantitativas():
  colunas = [
        'HP', 'Attack', 'Defense', 'Sp.Attack', 'Sp.Defense', 'Speed', 'Base_Stats',
        'normal_weakness', 'fire_weakness', 'water_weakness', 'electric_weakness',
        'grass_weakness', 'ice_weakness', 'fighting_weakness', 'poison_weakness',
        'ground_weakness', 'flying_weakness', 'psychic_weakness', 'bug_weakness',
        'rock_weakness', 'ghost_weakness', 'dragon_weakness', 'dark_weakness',
        'steel_weakness', 'fairy_weakness', 'number_immune', 'number_not_effective',
        'number_normal', 'number_super_effective'
    ]
  exibir_colunas(colunas, colunas_por_linha=4)
```

# Qual é a distribuição dos pontos de ataque (Attack) entre todos os Pokémon?
→ Média, mediana, desvio padrão, mínimo e máximo. Visualização dos dados com um histograma ou boxplot.

```python
attack = df['Attack']         # Tabela Attack
plt.figure(figsize=(10, 3))    # Tamanho da figura

# Estatísticas básicas
media = attack.mean()
mediana = attack.median()
desvio_padrao = attack.std()
minimo = attack.min()
maximo = attack.max()

print(f'Média: {media:.2f}')
print(f'Mediana: {mediana}')
print(f'Desvio Padrão: {desvio_padrao:.2f}')
print(f'Mínimo: {minimo}')
print(f'Máximo: {maximo}\n')

# Histograma
plt.subplot(1, 2, 1)
sns.histplot(attack, bins=20, kde=True, color='skyblue')
plt.title('Histograma')
plt.xlabel('Ataque')
plt.ylabel('Frequência')

# Boxplot
plt.subplot(1, 2, 2)
sns.boxplot(x=attack, color='lightgreen')
plt.title('Boxplot')
plt.xlabel('Ataque');   # ; para remover a saída "Text(0.5, 0, 'Ataque')"
```

# Como variam os pontos de vida (HP) entre as diferentes gerações de Pokémon?

→ Agrupe por geração (Generation) e apresente a média e desvio padrão do HP para cada grupo.

```python
media_desvio_HP = df.groupby('gen')["HP"].agg(MediaHP = 'mean', Desvio_PadraoHP = 'std').round(2) # Cria a Tebela

print("(Por Geração)\n\033[1mHP: Média   Desvio Padrão\n")
for idx, row in media_desvio_HP.iterrows():
    print(f"{idx}: \033[31m{row['MediaHP']:>6.2f}\033[0m  |  \033[34m{row['Desvio_PadraoHP']:>6.2f}\033[0m")
print()

plt.figure(figsize=(7, 3))
sns.barplot(data=media_desvio_HP, x='gen', y='MediaHP', color='red', legend=False)

# Personalização
plt.title('Média e Desvio Padrão de HP por Geração');
plt.xlabel('Geração')
plt.ylabel('HP Médio')
plt.xticks(rotation=45)

# Exibir o gráfico
plt.tight_layout()
plt.show()
```

# Quais são os Pokémon com as maiores e menores velocidades (Speed) e como se comporta essa variável na base como um todo?

→ Apresente as estatísticas descritivas completas da velocidade, incluindo o nome dos 5 Pokémon mais rápidos e dos 5 mais lentos.

```python
rapidos = df[['ID', 'Name', 'Speed']].sort_values(by='Speed', ascending=False).head(5) # ordenados pela velocidade
print("Os 5 mais rapidos:\n")
print(rapidos.to_string(index=False),"\n")

lentos = df[['ID', 'Name', 'Speed']].sort_values(by='Speed', ascending=True).head(5)    # a mesma coisa do anterior, só que de forma crescente
print("Os 5 mais lentos:\n")
print(lentos.to_string(index=False), "\n")

print("Estatísticas descritivas:\n")
est = df["Speed"].describe()  #estatísticas descritivas

estatísticas = pd.DataFrame({       #cria um data frame com objetivo de alterar individualmente cada parametro do describe()
    'Estatistíca': ['Quantidade', 'Média', 'Desvio Padrão', 'Mínimo', '1º Quartil', 'Mediana', '3º Quartil', 'Máximo'],
    'Valor': [
        int(est['count']),
        round(est['mean'], 2),
        round(est['std'], 2),
        int(est['min']),
        int(est['25%']),
        int(est['50%']),
        int(est['75%']),
        int(est['max'])
    ]
})

print(estatísticas.to_string(index=False))
```
